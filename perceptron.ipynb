```python
#importing important libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
#converting csv file to a dataframe
df=pd.read_csv("Single Layer Perceptron Dataset.csv")
```

```python
#viewing the data
df
```

```python
#viewing the features that are needed
df=df.head(13)
df
```

```python
#isolating the features
X=df.iloc[:,2:4]
y=df.iloc[:,4]
#changed y's datatype from float to int to plot mlxtend
y=y.astype(int)
```

```python
plt.scatter(x=df['Feature2'],y=df['Feature3'],c=df['Class_Label'])
```

```python
#importing perceptron
from sklearn.linear_model import Perceptron
p=Perceptron()
```

```python
#training the perceptron
p.fit(X,y)
```

```python
#values of weights
p.coef_
```

```python
#value of the bias
p.intercept_
```

```python
#plotting the decision region using mlxtend
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X.to_numpy(),y.to_numpy(),clf=p,legend=2)
```
